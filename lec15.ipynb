# Feature Scaling

```python
import numpy as np 
import pandas as pd 

from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, MaxAbsScaler

np.random.seed(42)
data = np.random.randint(0,100, size=(100,4))

df = pd.DataFrame(data, columns=["Feature 1", "Feature 2", "Feature 3", "Feature 4"])
df["Feature 5"] = np.random.randint(0,1000, size=(100,1))

min_max_scaler = MinMaxScaler()
df_min_max_scaled = pd.DataFrame(min_max_scaler.fit_transform(df), columns=df.columns)

standard_scaler = StandardScaler()
df_standard_scaled = pd.DataFrame(standard_scaler.fit_transform(df), columns=df.columns)

robust_scaler = RobustScaler()
df_robust_scaled = pd.DataFrame(robust_scaler.fit_transform(df), columns=df.columns)

max_abs_scaler = MaxAbsScaler()
df_max_abs_scaled = pd.DataFrame(max_abs_scaler.fit_transform(df), columns=df.columns)

print("Original Data")
print(df.head())

print("\nMin-Max Scaled Data")
print(df_min_max_scaled.head())

print("\nStandard Scaled Data")
print(df_standard_scaled.head())

print("\nRobust Scaled Data")
print(df_robust_scaled.head())

print("\nMax Abs Scaled Data")
print(df_max_abs_scaled.head())
```

# Feature Encoding 

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder

data = {
    'Colour': ['Red', 'Green', 'Blue',  'Blue', 'Red', 'Green'],
    'Size': ['M', 'L', 'S', 'XL', 'M', 'L'],
    'Price': [10, 15, 20, 25, 30, 35]
}

df = pd.DataFrame(data)

label_encoder = LabelEncoder()
df['Colour_LabelEncoded'] = label_encoder.fit_transform(df['Colour'])

one_hot_encoder = OneHotEncoder()
df_one_hot_encoded = pd.DataFrame(one_hot_encoder.fit_transform(df[['Colour']]).toarray(), columns=one_hot_encoder.get_feature_names_out(['Colour']))

ordinal_encoder = OrdinalEncoder()
df['Size_OrdinalEncoded'] = ordinal_encoder.fit_transform(df[['Size']])
print("Original Data")
print(df.head())

print("\nLabel Encoded Data")
print(df['Colour_LabelEncoded'])

print("\nOne Hot Encoded Data")
print(df_one_hot_encoded)

print("\nOrdinal Encoded Data")
print(df['Size_OrdinalEncoded'])
```

# Feature Engineering 
- Extracting / Transforming raw data into some meaningful features that improves the performance of ML models.
- It involves selecting / creating or modifying variables ( features / object features ) to better represent the underlying patterns in the data.


```python
import pandas as pd
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix

df = pd.read_csv("datasets/synthetic_dataset.csv")
# Fix any accidental trailing spaces in column names
df.columns=df.columns.str.strip()
# Seperate Feature and target
x=df.drop(columns=["target"])
y=df["target"]
# Split into training and testing sets
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
# Initialize and train the RandomForestClassifier
clf=RandomForestClassifier(n_estimators=100,random_state=42)
clf.fit(X_train,y_train)
```

```python
# Make Presictions
y_pred=clf.predict(X_test)
# Evaluate model performance
accuracy=accuracy_score(y_test,y_pred)
conf_matrix=confusion_matrix(y_test,y_pred)
report=classification_report(y_test,y_pred)
# Print results
print(f"Accuracy:{accuracy:.4f}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report")
print(report)
```

```python
# Seperate Feature and target
feature_columns=[col for col in df.columns if col.startswith("feature_")]
X=df[feature_columns]
y=df["target"]
# Compute statistical features for each row
X["mean"]=X.mean(axis=1)
X["std"]=X.std(axis=1)
X["min"]=X.min(axis=1)
X["max"]=X.max(axis=1)
X["median"]=X.median(axis=1)
X["sum"]=X.sum(axis=1)
X["variance"]=X.var(axis=1)
X["skew"]=X.skew(axis=1)
X["kurtosis"]=X.kurtosis(axis=1)

# Split into training and testing sets
X_train,X_test,y_train,y_test=train_test_split(X.iloc[:,-7:],y,test_size=0.2,random_state=42,stratify=y)
print(X_train.head())
```

```python
# Initialize and train the RandomForestClassifier
clf=RandomForestClassifier(n_estimators=100,random_state=42)
clf.fit(X_train,y_train)
```

```python
# Make Predictions
y_pred=clf.predict(X_test)
# Evaluate model performance
accuracy=accuracy_score(y_test,y_pred)
conf_matrix=confusion_matrix(y_test,y_pred)
report=classification_report(y_test,y_pred)
# Print results
print(f"Accuracy:{accuracy:.4f}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report")
print(report)

# Read t test for feature importance
```
